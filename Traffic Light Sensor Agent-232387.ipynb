```python
class TrafficLightAgent:
  
    
    def __init__(self):
        self.light_sequence = ["RED", "YELLOW", "GREEN"]
        print("Traffic Light Agent Initialized")
    
    def light_control(self, traffic_density, emergency_vehicle, pedestrian_crossing, weather="clear", sensor_error=False):
  
        if sensor_error:
            return "FLASH_RED_MODE"
        
  
        if emergency_vehicle:
            return "ALL_RED_EMERGENCY"
        

        if pedestrian_crossing:
            return "RED_FOR_PEDESTRIANS"
        
   
        if weather == "foggy":
            return "ACTIVATE_FOG_MODE"
        elif weather == "rainy":
            return "EXTEND_YELLOW_LIGHT"
        
 
        if traffic_density == "high":
            return "EXTEND_GREEN_LIGHT"
        elif traffic_density == "medium":
            return "NORMAL_CYCLE"
        elif traffic_density == "low":
            return "NIGHT_MODE_BLINKING"
        else:
            return "MAINTAIN_CURRENT"


def test_traffic_light_agent():
    traffic_agent = TrafficLightAgent()

    test_scenarios = [
        ("high", False, False, "clear", False, "EXTEND_GREEN_LIGHT"),    # normal
        ("medium", False, False, "clear", False, "NORMAL_CYCLE"),        # normal
        ("low", False, False, "clear", False, "NIGHT_MODE_BLINKING"),    # normal
        ("high", True, False, "clear", False, "ALL_RED_EMERGENCY"),      # emergency
        ("medium", False, True, "clear", False, "RED_FOR_PEDESTRIANS"),  # pedestrian
        ("low", True, True, "clear", False, "ALL_RED_EMERGENCY"),        # emergency priority
        ("medium", False, False, "foggy", False, "ACTIVATE_FOG_MODE"),   # foggy condition
        ("medium", False, False, "rainy", False, "EXTEND_YELLOW_LIGHT"), # rainy condition
        ("high", False, False, "clear", True, "FLASH_RED_MODE"),         # sensor error
    ]
    
    print("\n Testing Traffic Light Agent (Extended Conditions) ")
    print("=" * 60)
    
    for i, (density, emergency, pedestrian, weather, sensor, expected) in enumerate(test_scenarios, 1):
        action = traffic_agent.light_control(density, emergency, pedestrian, weather, sensor)
        status = "✔️" if action == expected else "❌"
        print(f"Test {i}:")
        print(f"   Density: {density:>6}, Emergency: {emergency:>5}, Pedestrian: {pedestrian:>5}, Weather: {weather:>6}, Sensor Error: {sensor}")
        print(f"   Action: {action} {status}\n")



test_traffic_light_agent()
```
