# 1. 🚪🚪🚪 Automatic Door Agent 🚪🚪🚪

```python
class AutomaticDoorAgent:
    """
    SIMPLE REFLEX AGENT FOR AUTOMATIC DOORS
    =======================================
    This agent makes decisions based SOLELY on current sensor inputs
    without any memory of past states or learning from history.
    
    Key Characteristics:
    - No internal state memory
    - Decisions based only on current percepts
    - Simple condition-action rules
    - Fast and predictable behavior
    """

    def __init__(self):
        """
        CONSTRUCTOR METHOD
        ==================
        Purpose: Initialize the agent when it's created
        Called automatically when we create: agent = AutomaticDoorAgent()
        """
        print("🚪🚪🚪 Automatic Door Agent Initialized 🚪🚪🚪")
        # Note: No instance variables needed for simple reflex agent
        # This demonstrates it doesn't maintain any internal state

    def door_control(self, motion_sensor, safety_sensor):
        """
        MAIN DECISION-MAKING METHOD
        ===========================
        This is where the simple reflex logic is implemented
        
        Parameters:
        - motion_sensor: Current state of motion detection
          Possible values: "detected" or "not_detected"
        - safety_sensor: Current state of safety check
          Possible values: "clear" or "obstacle"
        
        Returns: Action to be taken by the door mechanism
        """
        
        # RULE 1: If motion detected AND path is clear → OPEN DOOR
        # --------------------------------------------------------
        if motion_sensor == "detected" and safety_sensor == "clear":
            return "OPEN_DOOR"
            # Why this order? Safety first - check clearance before opening
        
        # RULE 2: If no motion detected → CLOSE DOOR  
        # ------------------------------------------
        elif motion_sensor == "not_detected":
            return "CLOSE_DOOR"
            # Simple reflex: No people = close door
            # Doesn't remember if someone just passed through
        
        # RULE 3: If obstacle detected → STOP AND REVERSE for safety
        # ----------------------------------------------------------
        elif safety_sensor == "obstacle":
            return "STOP_AND_REVERSE"
            # Safety override - prevents door from closing on objects/people
        
        # DEFAULT RULE: If none of above conditions match
        # -----------------------------------------------
        else:
            return "MAINTAIN_CURRENT_STATE"
            # Fallback action to prevent unexpected behavior

# TESTING FUNCTION
def test_door_agent():
    """
    COMPREHENSIVE TEST FUNCTION
    ===========================
    Purpose: Verify the agent works correctly in different scenarios
    Tests all possible combinations of sensor inputs
    """
    
    # STEP 1: Create an instance of the agent
    door_agent = AutomaticDoorAgent()
    # This calls __init__() and shows initialization message
    
    # STEP 2: Define test scenarios
    # Each scenario is a tuple: (motion_sensor, safety_sensor, expected_action)
    test_scenarios = [
        # Scenario 1: Normal entry - person approaching, path clear
        ("detected", "clear", "OPEN_DOOR"),
        
        # Scenario 2: No one there - close the door
        ("not_detected", "clear", "CLOSE_DOOR"),
        
        # Scenario 3: Safety issue - person/object in doorway
        ("detected", "obstacle", "STOP_AND_REVERSE"),
        
        # Scenario 4: No motion but obstacle (maybe object left in doorway)
        ("not_detected", "obstacle", "CLOSE_DOOR")
        # Note: This might seem counterintuitive but follows simple reflex rules
        # The agent only sees "no motion" and acts on that alone
    ]
    
    # STEP 3: Display test header
    print("\n🧪🧪 🧪  Testing Automatic Door Agent: 🧪 🧪 🧪 ")
    print("=" * 50)
    
    # STEP 4: Execute each test scenario
    for i, (motion, safety, expected) in enumerate(test_scenarios, 1):
        # i: test number (starts from 1 due to enumerate(..., 1))
        # motion, safety: input parameters for the agent
        # expected: what we think the agent should return
        
        # Call the agent's decision-making function
        action = door_agent.door_control(motion, safety)
        
        # Check if agent's action matches our expectation
        status = "✅" if action == expected else "❌"
        
        # Display test results
        print(f"Test {i}: Motion={motion}, Safety={safety}")
        print(f"       Action: {action} {status}")
        print()  # Empty line for readability

# EXECUTION BLOCK
# ===============
# This part runs when the script is executed directly
if __name__ == "__main__":
    test_door_agent()
```

# 2. 🌡️🌡️🌡️ Room Heater (Thermostat) Agent 🌡️🌡️🌡️

```python
class ThermostatAgent:
    """
    Simple reflex agent for room temperature control
    Uses current temperature reading only
    """
    
    def __init__(self, heating_threshold=18, cooling_threshold=26):
        self.heating_threshold = heating_threshold
        self.cooling_threshold = cooling_threshold
        print("🌡️🌡️ 🌡️   Thermostat Agent Initialized 🌡️ 🌡️ 🌡️ ")
        print(f"   Heating below: {heating_threshold}°C")
        print(f"   Cooling above: {cooling_threshold}°C")
    
    def temperature_control(self, current_temperature):
        """
        current_temperature: numeric value in Celsius
        """
        # Simple threshold-based rules
        if current_temperature < self.heating_threshold:
            return "TURN_ON_HEATER"
        elif current_temperature > self.cooling_threshold:
            return "TURN_ON_COOLER"
        else:
            return "MAINTAIN_TEMPERATURE"

# Lab Demonstration
def test_thermostat_agent():
    thermostat = ThermostatAgent(heating_threshold=18, cooling_threshold=26)
    
    # Test cases with different temperatures
    test_temperatures = [15, 20, 28, 18, 26, 22]
    
    print("\n🧪🧪🧪 Testing Thermostat Agent: 🧪🧪🧪")
    print("=" * 50)
    
    for temp in test_temperatures:
        action = thermostat.temperature_control(temp)
        print(f"Temperature: {temp:2d}°C → Action: {action}")
    
    print("\n📊📊📊 Temperature Control Summary:📊📊📊")
    print("❄️  Below 18°C: Heating")
    print("🌤️  18-26°C:    Maintain")
    print("🔥 Above 26°C:  Cooling")

# Run the test
test_thermostat_agent()
```

# 3. 🤖🤖🤖 Robot Vacuum (Basic Mode) Agent🤖🤖🤖

```python
class RobotVacuumAgent:
    """
    Simple reflex agent for basic robot vacuum
    Only considers current location and dirt status
    """
    
    def __init__(self):
        self.locations = ["A", "B", "C", "D"]
        print("🤖🤖🤖 Robot Vacuum Agent Initialized 🤖🤖🤖")
        print(f"   Monitoring locations: {self.locations}")
    
    def vacuum_decision(self, current_location, dirt_status, obstacle_detected):
        """
        current_location: "A", "B", "C", or "D"
        dirt_status: "dirty" or "clean"
        obstacle_detected: True or False
        """
        # Simple reflex rules
        if obstacle_detected:
            return "CHANGE_DIRECTION"
        elif dirt_status == "dirty":
            return f"CLEAN_{current_location}"
        else:
            # Move to next location in sequence
            current_index = self.locations.index(current_location)
            next_index = (current_index + 1) % len(self.locations)
            next_location = self.locations[next_index]
            return f"MOVE_TO_{next_location}"

# Lab Demonstration
def test_vacuum_agent():
    vacuum = RobotVacuumAgent()
    
    # Test scenarios: (location, dirt_status, obstacle, expected_action_pattern)
    test_scenarios = [
        ("A", "dirty", False, "CLEAN"),
        ("A", "clean", False, "MOVE_TO"),
        ("B", "dirty", True, "CHANGE_DIRECTION"),
        ("C", "clean", False, "MOVE_TO"),
        ("D", "dirty", False, "CLEAN")
    ]
    
    print("\n🧪🧪🧪 Testing Robot Vacuum Agent:🧪🧪🧪")
    print("=" * 50)
    
    for location, dirt, obstacle, expected in test_scenarios:
        action = vacuum.vacuum_decision(location, dirt, obstacle)
        status = "✅" if expected in action else "❌"
        print(f"Location: {location}, Dirt: {dirt}, Obstacle: {obstacle}")
        print(f"Action: {action} {status}")
        print()

# Run the test
test_vacuum_agent()
```

# 4. 🚦🚦🚦 Traffic Light Sensor Agent 🚦🚦🚦

```python
class TrafficLightAgent:
    """
    Simple reflex agent for traffic light control
    Uses current traffic density and emergency vehicle detection
    """
    
    def __init__(self):
        self.light_sequence = ["RED", "YELLOW", "GREEN"]
        print("🚦🚦🚦 Traffic Light Agent Initialized 🚦🚦🚦")
    
    def light_control(self, traffic_density, emergency_vehicle, pedestrian_crossing):
        """
        traffic_density: "low", "medium", "high"
        emergency_vehicle: True or False
        pedestrian_crossing: True or False
        """
        # Priority 1: Emergency vehicles
        if emergency_vehicle:
            return "ALL_RED_EMERGENCY"
        
        # Priority 2: Pedestrian safety
        if pedestrian_crossing:
            return "RED_FOR_PEDESTRIANS"
        
        # Normal operation based on traffic density
        if traffic_density == "high":
            return "EXTEND_GREEN_LIGHT"
        elif traffic_density == "medium":
            return "NORMAL_CYCLE"
        elif traffic_density == "low":
            return "NIGHT_MODE_BLINKING"
        else:
            return "MAINTAIN_CURRENT"

# Lab Demonstration
def test_traffic_light_agent():
    traffic_agent = TrafficLightAgent()
    
    # Test scenarios
    test_scenarios = [
        ("high", False, False, "EXTEND_GREEN_LIGHT"),
        ("medium", False, False, "NORMAL_CYCLE"),
        ("low", False, False, "NIGHT_MODE_BLINKING"),
        ("high", True, False, "ALL_RED_EMERGENCY"),
        ("medium", False, True, "RED_FOR_PEDESTRIANS"),
        ("low", True, True, "ALL_RED_EMERGENCY")  # Emergency has highest priority
    ]
    
    print("\n🧪🧪🧪 Testing Traffic Light Agent:🧪🧪🧪")
    print("=" * 50)
    
    for i, (density, emergency, pedestrian, expected) in enumerate(test_scenarios, 1):
        action = traffic_agent.light_control(density, emergency, pedestrian)
        status = "✅" if action == expected else "❌"
        
        print(f"Test {i}:")
        print(f"   Density: {density:>6}, Emergency: {emergency:>5}, Pedestrian: {pedestrian:>5}")
        print(f"   Action: {action} {status}")
        print()

# Run the test
test_traffic_light_agent()
```
