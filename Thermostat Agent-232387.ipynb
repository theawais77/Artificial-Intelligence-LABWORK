```python
class ThermostatAgent:
 
    def __init__(self, heating_threshold=18, cooling_threshold=26):
        self.heating_threshold = heating_threshold
        self.cooling_threshold = cooling_threshold
        print("Thermostat Agent Initialized ")
        print(f"   Heating below: {heating_threshold}°C")
        print(f"   Cooling above: {cooling_threshold}°C")
    
    def temperature_control(self, current_temperature):
        
        if current_temperature < 5:
            return "FREEZING_WARNING"

        elif current_temperature > 35:
            return "OVERHEAT_WARNING"
        
        elif current_temperature < self.heating_threshold:
            return "TURN_ON_HEATER"
        elif current_temperature > self.cooling_threshold:
            return "TURN_ON_COOLER"
        elif current_temperature == self.heating_threshold:
            return "PREPARE_TO_HEAT"
        elif current_temperature == self.cooling_threshold:
            return "PREPARE_TO_COOL"
        elif current_temperature == (self.heating_threshold + self.cooling_threshold) // 2:
            return "OPTIMAL_TEMPERATURE"
        else:
            return "MAINTAIN_TEMPERATURE"


def test_thermostat_agent():
    thermostat = ThermostatAgent(heating_threshold=18, cooling_threshold=26)
  
    test_temperatures = [15, 18, 20, 22, 26, 28,36,4]
    
    print("\n Testing Thermostat Agent: ")
    print("=" * 50)
    
    for temp in test_temperatures:
        action = thermostat.temperature_control(temp)
        print(f"Temperature: {temp:2d}°C → Action: {action}")
    
    print("\n Temperature Control Summary:")
    print("Below 5°C:  Freezing Warning")
    print("Below 18°C: Heating")
    print("  Exactly 18°C: Prepare to Heat")
    print("18-26°C:    Maintain")
    print("  Exactly 22°C: Optimal Temperature")
    print("Exactly 26°C: Prepare to Cool")
    print(" Above 26°C:  Cooling")
    print(" Above 35°C:  Overheat Warning")

test_thermostat_agent()
```
