```python
class RobotVacuumAgent:
    
    def __init__(self):
        self.locations = ["A", "B", "C", "D"]
        print(" Robot Vacuum Agent Initialized ")
        print(f"   Monitoring locations: {self.locations}")
    
    def vacuum_decision(self, current_location, dirt_status, obstacle_detected):
  
        if obstacle_detected:
            return "CHANGE_DIRECTION"
        
        # Handle different types of dirt
        elif dirt_status == "heavy":
            return f"DEEP_CLEAN_{current_location}"
        elif dirt_status == "medium":
            return f"NORMAL_CLEAN_{current_location}"
        elif dirt_status == "light":
            return f"QUICK_CLEAN_{current_location}"
        elif dirt_status == "dirty": 
            return f"CLEAN_{current_location}"
        else:
         
            current_index = self.locations.index(current_location)
            next_index = (current_index + 1) % len(self.locations)
            next_location = self.locations[next_index]
            return f"MOVE_TO_{next_location}"

def test_vacuum_agent():
    vacuum = RobotVacuumAgent()
    
    test_scenarios = [
        ("A", "light", False, "QUICK_CLEAN"),
        ("A", "medium", False, "NORMAL_CLEAN"),
        ("B", "heavy", False, "DEEP_CLEAN"),
        ("C", "clean", False, "MOVE_TO"),
        ("D", "dirty", False, "CLEAN"), 
        ("C", "medium", True, "CHANGE_DIRECTION"),  
    ]
    
    print("\n Testing Robot Vacuum Agent (Dirt Types)")
    print("=" * 60)
    
    for location, dirt, obstacle, expected in test_scenarios:
        action = vacuum.vacuum_decision(location, dirt, obstacle)
        status = "✔️" if expected in action else "❌"
        print(f"Location: {location}, Dirt: {dirt}, Obstacle: {obstacle}")
        print(f"Action: {action} {status}\n")


test_vacuum_agent()
```
