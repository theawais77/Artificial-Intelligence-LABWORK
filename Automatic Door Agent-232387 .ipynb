# 1. Automatic Door Agent 

```python
class AutomaticDoorAgent:
    def __init__(self):
        print(" Automatic Door Agent Initialized ")
   
    def door_control(self, motion_sensor, safety_sensor):
        if motion_sensor == "detected" and safety_sensor == "clear":
            return "OPEN_DOOR"
        elif motion_sensor == "not_detected":
            return "CLOSE_DOOR"
        elif safety_sensor == "obstacle":
            return "STOP_AND_REVERSE"
        elif motion_sensor == "detected" and safety_sensor == "warning":
            return "SLOW_OPEN"
        elif motion_sensor == "detected" and safety_sensor == "low_battery":
            return "ALERT_LOW_BATTERY"
        elif motion_sensor == "unknown":
            return "CHECK_SENSORS"
        else:
            return "MAINTAIN_CURRENT_STATE"
          
def test_door_agent():
    door_agent = AutomaticDoorAgent()
   
    test_scenarios = [
        ("detected", "clear", "OPEN_DOOR"),
        ("not_detected", "clear", "CLOSE_DOOR"),
        ("detected", "obstacle", "STOP_AND_REVERSE"),
        ("not_detected", "obstacle", "CLOSE_DOOR"),
        ("detected", "warning", "SLOW_OPEN"),
        ("detected", "low_battery", "ALERT_LOW_BATTERY"),
        ("unknown", "clear", "CHECK_SENSORS"),
    ]
 
    print("\n  Testing Automatic Door Agent:  ")
    print("=" * 50)
    
    for i, (motion, safety, expected) in enumerate(test_scenarios, 1):
        action = door_agent.door_control(motion, safety)
        status = "✔️" if action == expected else "❌"
        print(f"Test {i}: Motion={motion}, Safety={safety}")
        print(f"       Action: {action} {status}")
        print()  

if __name__ == "__main__":
    test_door_agent()
```
